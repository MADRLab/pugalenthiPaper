```python
import sys
import os
from multilingual_clip import pt_multilingual_clip
import transformers
import open_clip
import torch
from PIL import Image
import clip
import torch.nn.functional as F

def mainImageTextSimScore(inputDirectory, imagePath,outputName):
  simScores = []
  for i, visionLangModel in enumerate(visionLangModels):
    print(f"Derive image-text congruence scores with visionLanguageModel: {modelPairNames[i]}")
    tokenizer = transformers.AutoTokenizer.from_pretrained(textEncoderName)
    simScores.append(simScore(visionLangModel, tokenizer, inputDirectory,imagePath))
  # Create a DataFrame from the collected data
  import pandas as pd
  mCLIPdf = pd.DataFrame(simScores).T
  mCLIPdf.columns = modelPairNames
  mCLIPdf.index = [file for file in sorted(os.listdir(inputDirectory))]
  mCLIPdf.head()
  mCLIPdf.to_csv(outputName + '_imageTextCongruenceScores.csv')
  from google.colab import files
  files.download(outputName + '_imageTextCongruenceScores.csv')

def simScore(visionLanguageModel, tokenizer, inputDirectory,imagePath):
    simScores = []
    # As image is constant, compute image features once before for loop
    imageEncoder, textEncoder, preprocess = visionLanguageModel
    image = preprocess(Image.open(imagePath)).unsqueeze(0).to(device)
    image_features = imageEncoder.encode_image(image).to(torch.float32).to(device)
    for file in sorted(os.listdir(inputDirectory)):
      f = open(f'{inputDirectory}/{file}', 'r')
      text = f.read()
      f.close()
      try:
        textEncoder= textEncoder.to('cpu')
        text_features = textEncoder.forward(text, tokenizer)
      except:
        cutoff = 1650
        print(f"File {file} has too many characters for text encoder: {len(text)}. Will clip to first {cutoff} characters")
        text = text[:cutoff]
        text_features = textEncoder.forward(text, tokenizer)
      text_features = text_features.to(torch.float32).to(device)
      #compute cosine similarity
      #print(text_features.shape, image_features.shape)
      final_similarity_score = F.cosine_similarity(text_features, image_features).item()
      #final_similarity_score = (text_features @ image_features.T).mean().item()
      simScores.append(final_similarity_score)
    return simScores
```
